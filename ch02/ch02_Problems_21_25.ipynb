#### 21.

page 57

A 50.0-g Super Ball traveling at 25.0 m/s bounces off
a brick wall and rebounds at 22.0 m/s. 
A high-speed camera records this event. If the ball is
in contact with the wall for 3.50 ms, what is the magnitude of
the average acceleration of the ball during this time interval?



```python
v0_qty = 25.0 * (u.m / u.s)
vf_qty = -22.0 * (u.m / u.s)
Delta_t_qty = (3.50 * u.ms).to(u.s)

# a = Delta_v / Delta_t
a_qty = (vf_qty - v0_qty) / Delta_t_qty

a_mag_qty = abs(a_qty)

print(f"Magnitude average acceleration: {a_mag_qty:.3g} equivalent to {a_mag_qty:,.1f}")
```

#### 22. 

The average person passes out at an acceleration of 7g
(that is, seven times the gravitational acceleration on Earth).
Suppose a car is designed to accelerate at this rate. How much
time would be required for the car to accelerate from rest to
60.0 miles per hour? (The car would need rocket boosters!)

```python
a_mag_qty = 7 * const.g0
vf_qty = 60.0 * (u.imperial.mi / u.h)
v0_qty = 0.0 * (u.imperial.mi / u.h)

# a = Delta_v / Delta_t ==> Delta_t = Delta_v / a
Delta_t_qty = (vf_qty - v0_qty) / a_mag_qty

print(f"Required time: {Delta_t_qty.decompose():.3g}")
```

### 23. 

A certain car is capable of accelerating at a rate of
0.60 m/s2. How long does it take for this car to go from a
speed of 55 mi/h to a speed of 60 mi/h?

```python
a_mag_qty = 0.60 * (u.m / u.s**2)
vf_qty = 60 * (u.imperial.mi / u.h)
v0_qty = 55 * (u.imperial.mi / u.h)

# a = Delta_v / Delta_t ==> Delta_t = Delta_v / a
Delta_t_qty = (vf_qty - v0_qty) / a_mag_qty

print(f"Required time: {Delta_t_qty.decompose():.2g}")
```

#### 25. 

HMS Queen Elizabeth is the largest warship ever built for the
Royal Navy. If its launcher propels an F-35 from rest to a speed
of 78.5 m/s in 2.50 s, determine 

* **(a)** the plane’s average acceleration and 

* **(b)** the distance the plane moves.

```python
vf_qty = 78.5 * (u.m / u.s)
v0_qty = 0 * (u.m / u.s)
Delta_t_qty = 2.50 * u.s

print("\n--- Part (a) ---")
a_qty = (vf_qty - v0_qty) / Delta_t_qty

print(f"Plane's average acceleration: {a_qty:.3g}")

print("\n--- Part (b) ---")
# x = v0 t + (1/2) a t^2 (v0=0)
x = (1 / 2) * a_qty * Delta_t_qty**2

print(f"Plane's distance: {x:.3g}")
```
