#### 1.

The speed of a nerve impulse in the human body is
about 100 m/s. If you accidentally stub your toe in the dark,
estimate the time it takes the nerve impulse to travel to your
brain.

```python
speed_qty = 100 * (u.m / u.s)
# Assumption length nerve from toe to brain 2 m
distance_qty = 2 * u.m
# v = d / t ==> t = d / v
t_qty = distance_qty / speed_qty
print(f"Estimated time to reach the brain: t ≈ {t_qty}")
```

#### 2. 

Light travels at a speed of about $3 \times 10^8 m/s$. 

* **(a)** How many miles does a pulse of light travel in a time interval of 0.1 s,
which is about the blink of an eye? 

* **(b)** Compare this distance to the diameter of Earth.

```python
speed_light_qty = 3e8 * (u.m / u.s)
dt_qty = 0.1 * u.s

print("\n--- Part (a) ---")
# v = dx / dt ==> dx = v dt
dx_qty = speed_light_qty * dt_qty

dx_mi_qty = dx_qty.to(u.imperial.mi)

print(
    f"In the interval of {dt_qty} a pulse of ligth travel: dx = {dx_mi_qty:,.0f} equivalent to: {dx_mi_qty:.2e} "
)

print("\n--- Part (b) ---")
ratio = dx_qty / (2 * const.R_earth)
print(f"This distance is about: {ratio:,.1f} times the diameter of Earth")
```

#### 3. 

A person travels by car from one city to another with different
constant speeds between pairs of cities. She drives for
30.0 min at 80.0 km/h, 12.0 min at 100 km/h, and 45.0 min at
40.0 km/h and spends 15.0 min eating lunch and buying gas.

* **(a)** Determine the average speed for the trip. 

* **(b)** Determine the distance between the initial and final cities along the route.

```python
# Define time intervals for each segment and the stop
t1 = 30.0 * u.min
v1 = 80.0 * u.km / u.h

t2 = 12.0 * u.min
v2 = 100.0 * u.km / u.h

t3 = 45.0 * u.min
v3 = 40.0 * u.km / u.h

t_break = 15.0 * u.min  # Lunch and gas break

# (b) Calculate total distance between cities
d1 = v1 * t1
d2 = v2 * t2
d3 = v3 * t3

total_distance = (d1 + d2 + d3).to(u.km)


# (a) Calculate average speed for the trip
total_time = (t1 + t2 + t3 + t_break).to(u.h)
average_speed = (total_distance / total_time).to(u.km / u.h)

print("\n--- Part (a) ---")
print(f"Total Time: {total_time:.2f}")
print(f"Average Speed: {average_speed:.1f}")

print("\n--- Part (b) ---")
print(f"Total Distance: {total_distance:.1f}")
```

#### 4. 
A rugby player runs 100 m from his own goal line to the
opposing team’s goal line, returning to the halfway line, all in
18.0 s. Calculate 

* **(a)** his average speed, and 

* **(b)** the magnitude of his average velocity.

```python
# distance
d1 = 100 * u.m
d2 = d1 / 2

# displacement
dsp = d1 - d2

# time
total_time = 18.0 * u.S

# (a) calculate average speed
total_distance = d1 + d2
average_speed = total_distance / total_time

# (b) calculate magnitude average velocity
average_vel_mag = dsp / total_time

print("\n--- Part (a) ---")
print(f"Average speed: {average_speed:.3g}")

print("\n--- Part (b) ---")
print(f"Magnitude average velocity: {average_vel_mag:.3g}")
```

#### 5. 
Two boats start together and race across a 60-km-wide lake and
back. Boat A goes across at 60 km/h and returns at 60 km/h.
Boat B goes across at 30 km/h, and its crew, realizing how far
behind it is getting, returns at 90 km/h. Turnaround times are
negligible, and the boat that completes the round trip first
wins. 

* **(a)** Which boat wins and by how much? (Or is it a tie?)

* **(b)** What is the average velocity
of the winning boat?

```python
d = 60 * u.km

# outbound journey
vel_outbound_A = 60 * (u.km / u.h)
vel_outbound_B = 30 * (u.km / u.h)

# return journey
vel_return_A = 60 * (u.km / u.h)
vel_return_B = 90 * (u.km / u.h)

# calculate time
# v = d / t ==> t = d/v
t_outbound_A = d / vel_outbound_A
t_return_A = d / vel_return_A
total_time_A = t_outbound_A + t_return_A

t_outbound_B = d / vel_outbound_B
t_return_B = d / vel_return_B
total_time_B = t_outbound_B + t_return_B

delta_t = total_time_B - total_time_A

print("\n--- Part (a) ---")
print(
    f"Outbound journey: boat A spends t={t_outbound_A}, boat B spends t={t_outbound_B}"
)
print(f"Return journey: boat A spends t={t_return_A}, boat B spends t={t_return_B:.2g}")
print(
    f"Total journey: boat A spends t={total_time_A}, boat B spends t={total_time_B:.2g}"
)

print(
    f"Total journey: boat A spends t={total_time_A}, boat B spends t={total_time_B:.2g}"
)
print(f"Boat A wins by t={delta_t:.2g}, distance = {d}")
print(f"When Boat A finished, Boat B just finished the outboud journey")

# (b) calculate magnitude average velocity
# displacement
dsp = d - d
average_vel_mag = dsp / total_time_A

print("\n--- Part (b) ---")
print(
    f"Total journey boat A, starts and finishes at the same place, so displacement = {dsp},"
)
print(f"so the average velocity magnitude is: {average_vel_mag}.")
```

#### 7. 

A motorist drives north for 35.0 minutes at 85.0 km/h and

130. km in 2.00 h.

length of the runway is 2.5 km, will the plane be able to use
this runway safely? Defend your answer.

```python
v_to_qty = 75 * (u.m / u.s)  # velocity takeoff
a_qty = 1.3 * (u.m / u.s**2)
dx_qty = (2.5 * u.km).to(u.m)

# calculate the distance when the plane reaches the takeoff speed
x, x0, v, v0, a = sp.symbols("x, x_0, v, v_0, a", real=True)
t = sp.symbols("t", real=True, positive=True)
# --- Velocity square eq ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * x)
print(f"\n--- Velocity square equation: ---")
display(Math(rf"{sp.latex(eq_vel_sq)}"))

print("\n --- Solve symbolically for x ---")
x_vel_expr = sp.solve(eq_vel_sq, x)[0]
display(Math(rf"x = {sp.latex(x_vel_expr)}"))


x_to_qty = (v_to_qty**2 - 0) / (2 * a_qty)
print(f"Distance plane travel before takeoff is x = {x_to_qty.to(u.km):.1f}")
print(f"So, yes the plane is safe to take this runway of {dx_qty.to(u.km):.1f}")


# print(f"The calculated velocity at the final runway is: v={v:.2g}, the plane has a velocity margin of: {v_margin:.2g}")
```

#### 10. 

Two cars travel in the same direction along a straight highway,
one at a constant speed of 55 mi/h and the other at 70 mi/h.

* **(a)** Assuming they start at the same point, how much sooner
does the faster car arrive at a destination 10 mi away? 

* **(b)** How far must the faster car travel before it has a 15-min lead on the
slower car?

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
v1, v2 = sp.symbols(
    "v_1 v_2", real=True, positive=True
)  # Speeds (v1 = slower, v2 = faster)
d_a = sp.symbols("d_a", real=True, positive=True)  # Distance for part (a)
delta_t_b = sp.symbols("Delta_t_b", real=True, positive=True)  # Lead time for part (b)
d_b = sp.symbols("d_b", real=True, positive=True)  # Unknown distance for part (b)

# --- Part (a) ---
# Time taken by slower car: t1 = d_a / v1
# Time taken by faster car: t2 = d_a / v2
# Difference in arrival time: delta_t_a = t1 - t2
delta_t_a_expr = sp.simplify((d_a / v1) - (d_a / v2))

# --- Part (b) ---
# Time difference as a function of distance d_b: delta_t_b = (d_b / v1) - (d_b / v2)
# Solve for d_b
eq_b = sp.Eq(delta_t_b, (d_b / v1) - (d_b / v2))
d_b_expr = sp.simplify(sp.solve(eq_b, d_b)[0])

# Render LaTeX formulas nicely in Jupyter
display(
    Math(
        rf"\text{{Part (a) Time difference: }} \Delta t_a = {sp.latex(delta_t_a_expr)}"
    )
)
display(Math(rf"\text{{Part (b) Required distance: }} d_b = {sp.latex(d_b_expr)}"))
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical functions
calc_delta_t_a = sp.lambdify((d_a, v1, v2), delta_t_a_expr, modules="numpy")
calc_d_b = sp.lambdify((delta_t_b, v1, v2), d_b_expr, modules="numpy")

# Instantiate quantities with AstroPy units
v1_qty = 55.0 * (u.imperial.mile / u.h)
v2_qty = 70.0 * (u.imperial.mile / u.h)
d_a_qty = 10.0 * u.imperial.mile
delta_t_b_qty = 15.0 * u.min

# Evaluate Part (a) directly with Quantity objects
delta_t_a_qty = calc_delta_t_a(d_a_qty, v1_qty, v2_qty)

# Evaluate Part (b) directly with Quantity objects
d_b_qty = calc_d_b(delta_t_b_qty, v1_qty, v2_qty)

# Print Results
print("--- Final Results ---")
print(
    f"(a) The faster car arrives {delta_t_a_qty:.2f} sooner (or ~{delta_t_a_qty.to(u.min):.1f}) (or ~{delta_t_a_qty.to(u.s):.1f})."
)
print(
    f"(b) The faster car must travel {d_b_qty.decompose().to(u.imperial.mi):.2f} to gain a 15-minute lead."
)
```
