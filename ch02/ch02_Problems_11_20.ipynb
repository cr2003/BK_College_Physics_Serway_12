#### 11.

The cheetah can reach a top speed of 114 km/h (71 mi/h).
While chasing its prey in a short sprint, a cheetah starts from
rest and runs 45 m in a straight line, reaching a final speed of
72 km/h. 

* **(a)** Determine the cheetah’s average acceleration during
the short sprint, and 

* **(b)** find its displacement at $t = 3.5 s$.

```python
x_qty = 45 * u.m
v_qty = (72 * (u.km / u.h)).to(u.m / u.s)
v0_qty = 0 * (u.m / u.s)
t_b_qty = 3.5 * u.s


print("\n --- Part (a) ---")
# eq: v^2 = v0^2 + 2ax ==> a = (v^2 - v0^2) / (2 * x)
a_qty = (v_qty**2 - v0_qty**2) / (2 * x_qty)

print(f"the cheetah's average acceleration: a = {a_qty:.2g}")
```

```python
print("\n --- Part (b) ---")

# x = x0 + v0 t + (1/2) a t^2
# as x0=0, v0=0 the equation is:
# x = (1/2) a t^2
x_qty = (1 / 2) * a_qty * t_b_qty**2
print(f"Displacement: x = {x_qty:.2g}")
```

#### 12. 

An athlete swims the length L of a pool in a time t 1 and
makes the return trip to the starting position in a time t 2. If
she is swimming initially in the positive x - direction, determine
her average velocities symbolically in 

* **(a)** the first half of the swim, 

* **(b)** the second half of the swim, and 

* **(c)** the round trip. 

* **(d)** What is her average speed for the round trip?

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
L, t1, t2 = sp.symbols("L t1 t2", real=True, positive=True)

# (a) Average velocity for the first half (displacement = +L, time = t1)
v_avg_a_expr = sp.simplify(L / t1)

# (b) Average velocity for the second half (displacement = -L, time = t2)
v_avg_b_expr = sp.simplify(-L / t2)

# (c) Average velocity for the round trip (total displacement = 0, total time = t1 + t2)
v_avg_c_expr = sp.simplify(0 / (t1 + t2))

# (d) Average speed for the round trip (total distance = 2L, total time = t1 + t2)
v_avg_d_expr = sp.simplify((2 * L) / (t1 + t2))

# Render symbolic results using LaTeX
display(
    Math(
        r"\text{(a) First half average velocity: } v_{1,\text{avg}} = "
        + sp.latex(v_avg_a_expr)
    )
)
display(
    Math(
        r"\text{(b) Second half average velocity: } v_{2,\text{avg}} = "
        + sp.latex(v_avg_b_expr)
    )
)
display(
    Math(
        r"\text{(c) Round trip average velocity: } v_{\text{round,avg}} = "
        + sp.latex(v_avg_c_expr)
    )
)
display(
    Math(
        r"\text{(d) Round trip average speed: } s_{\text{avg}} = "
        + sp.latex(v_avg_d_expr)
    )
)
```

#### 13. 

A person takes a trip, driving with a constant speed of
89.5 km/h, except for a 22.0-min rest stop. If the person’s
average speed is 77.8 km/h, 

* **(a)** how much time is spent on
the trip and 

* **(b)** how far does the person travel?

```python
v, v_av, t1, t2 = sp.symbols("v, v_av, t1, t2", real=True, positive=True)

eq_v_av = sp.Eq(v_av, (v * t1 + 0) / (t1 + t2))
t1_expr = sp.solve(eq_v_av, t1)[0]

display(Math(r"\text{Driven part of the journey: } t_{1} = " + sp.latex(t1_expr)))
```

```python
# --- 1. Derive callables from SymPy ---
calc_t1 = sp.lambdify((t2, v_av, v), t1_expr)

v_qty = 89.5 * (u.km / u.h)
v_av_qty = 77.8 * (u.km / u.h)
t_rest_qty = (22.0 * u.min).to(u.h)

t1_qty = calc_t1(t_rest_qty, v_av_qty, v_qty)

print("\n --- Part (a) ---")
print(f"Time driven: {t1_qty:.3g}")
total_time_qty = t1_qty + t_rest_qty
print(f"Total time for the trip: {total_time_qty:.3g} ")

# eq: v = x / t ==> x = v * t
displacement_qty = v_qty * t1_qty

print("\n --- Part (b) ---")
print(f"Total displacement for the trip: {displacement_qty:.3g} ")
```

#### 14

A tortoise can run with a speed of 0.10 m/s, and a hare can
run 20 times as fast. In a race, they both start at the same time,
but the hare stops to rest for 2.0 minutes. The tortoise wins by
a shell (20 cm). 

* **(a)** How long does the race take? 

* **(b)** What is the length of the race?

```python
t, x, v_tourtoise, v_hare, t_rest, delta_x = sp.symbols(
    "t, x, v_tourtoise, v_hare, t_rest, Delta_x ", real=True, positive=True
)

# race tortoise
x_tourtoise = v_tourtoise * t
# race hare
x_hare = v_hare * (t - t_rest) + delta_x

eq_x = sp.Eq(x_tourtoise, x_hare)
# Solve for t
t_expr = sp.solve(eq_x, t)[0]

display(Math(r"\text{Equation of the race: }" + sp.latex(eq_x)))

display(Math(r"\text{Solving for t: }" + sp.latex(t_expr)))
```

```python
# --- 1. Derive callables from SymPy ---
calc_t = sp.lambdify((delta_x, t_rest, v_hare, v_tourtoise), t_expr)

# --- AstroPy vars
v_tourtoise_qty = 0.10 * (u.m / u.s)
v_hare_qty = 20 * v_tourtoise_qty
t_rest_qty = (2.0 * u.min).to(u.s)
delta_x_qty = (20 * u.cm).to(u.m)

t_qty = calc_t(delta_x_qty, t_rest_qty, v_hare_qty, v_tourtoise_qty)

print("\n --- Part (a) ---")
print(f"The tortoise reaches the finish line at time {t_qty:.1f}")


print("\n --- Part (b) ---")
# v = x / t ==> x = v t
x_qty = v_tourtoise_qty * t_qty
print(f"The length of the race is {x_qty:.2f}")
```

#### 15. 

To qualify for the finals in a racing event, a race car must
achieve an average speed of 250. km/h on a track with a total
length of $1.60 \times 10^3$. If a particular car covers the first half of
the track at an average speed of 230. km/h, what minimum
average speed must it have in the second half of the event to
qualify?

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
v_target = sp.symbols(
    "v_target", real=True, positive=True
)  # Required overall average speed (250 km/h)
v1 = sp.symbols("v1", real=True, positive=True)  # First half average speed (230 km/h)
v2 = sp.symbols("v2", real=True, positive=True)  # Second half average speed (unknown)
L = sp.symbols("L", real=True, positive=True)  # Total track length (1.60 x 10^3 m)

# --- Physical Relations ---
# Time taken for first half:  t1 = (L / 2) / v1
# Time taken for second half: t2 = (L / 2) / v2
# Total time: t_total = t1 + t2
# Target average speed: v_target = L / t_total = L / ((L / 2) * (1/v1 + 1/v2)) = 2 / (1/v1 + 1/v2)

eq = sp.Eq(v_target, 2 / ((1 / v1) + (1 / v2)))

# Solve for required speed in the second half (v2)
v2_expr = sp.simplify(sp.solve(eq, v2)[0])

# Render symbolic result using LaTeX (using raw string concatenation)
display(Math(r"\text{Required second-half average speed: } v_2 = " + sp.latex(v2_expr)))
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical function
calc_v2 = sp.lambdify((v_target, v1), v2_expr, modules="numpy")

# Instantiate quantities with AstroPy units
v_target_qty = 250.0 * (u.km / u.h)
v1_qty = 230.0 * (u.km / u.h)
L_qty = (
    1.60e3 * u.m
)  # Provided track length (note: v2 is independent of track length L)

# Evaluate result directly with Quantity objects
v2_qty = calc_v2(v_target_qty, v1_qty)

# Print Results
print("--- Final Results ---")
print(
    f"Minimum average speed in second half: {v2_qty:.2f} (or ~{v2_qty.to(u.m / u.s):.2f})"
)
```

#### 16. 

A paper in the journal Current Biology tells of some
jellyfish-like animals that attack their prey by launching stinging
cells in one of the animal kingdom’s fastest movements.
High-speed photography showed the cells were accelerated
from rest for 700. ns at $5.30 \times 10^7 m/s^2$. Calculate 

* **(a)** the maximum speed reached by the cells and 

* **(b)** the distance
traveled during the acceleration.

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
a = sp.symbols("a", real=True, positive=True)  # Acceleration
t = sp.symbols("t", real=True, positive=True)  # Acceleration duration
v_0 = sp.symbols("v_0", real=True, nonnegative=True)  # Initial speed (rest = 0)

# (a) Maximum speed reached starting from rest (v = v_0 + a * t)
v_max_expr = sp.simplify(v_0 + a * t)

# (b) Distance traveled during acceleration starting from rest (d = v_0 * t + 1/2 * a * t^2)
d_expr = sp.simplify(v_0 * t + sp.Rational(1, 2) * a * t**2)

# Render symbolic results using LaTeX
display(Math(r"\text{(a) Maximum speed: } v_{\text{max}} = " + sp.latex(v_max_expr)))
display(Math(r"\text{(b) Distance traveled: } d = " + sp.latex(d_expr)))
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical functions
calc_v_max = sp.lambdify((a, t, v_0), v_max_expr, modules="numpy")
calc_d = sp.lambdify((a, t, v_0), d_expr, modules="numpy")

# Instantiate quantities with AstroPy units
a_qty = 5.30e7 * (u.m / (u.s**2))  # Acceleration: 5.30 x 10^7 m/s^2
t_qty = 700.0 * u.ns  # Duration: 700 ns
v_0_qty = 0.0 * (u.m / u.s)  # Accelerated from rest

# Evaluate results directly with Quantity objects
v_max_qty = calc_v_max(a_qty, t_qty, v_0_qty)
d_qty = calc_d(a_qty, t_qty, v_0_qty)

# Print Results
print("--- Final Results ---")
print(f"(a) Maximum speed reached : {v_max_qty.to(u.m / u.s):.2f}")
print(f"(b) Distance traveled    : {d_qty.to(u.mm):.3f} (or ~{d_qty.to(u.um):.1f})")
```

#### 19. 

Runner A is initially 4.0 mi west of a flagpole and is running
with a constant velocity of 6.0 mi/h due east. Runner B is initially
3.0 mi east of the flagpole and is running with a constant
velocity of 5.0 mi/h due west. How far are the runners from
the flagpole when they meet?

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
# Flagpole is set at origin x = 0. East is positive (+x), West is negative (-x).
x0_A, x0_B = sp.symbols("x0_A x0_B", real=True)  # Initial positions
v_A, v_B = sp.symbols("v_A v_B", real=True)  # Constant velocities
t_meet, x_meet = sp.symbols(
    "t_meet x_meet", real=True
)  # Meeting time and meeting position
d_flagpole = sp.symbols(
    "d_flagpole", real=True, positive=True
)  # Distance from flagpole when meeting

# Position equations as functions of time t
# x_A(t) = x0_A + v_A * t
# x_B(t) = x0_B + v_B * t

# Set positions equal to find meeting time t_meet
eq_meet = sp.Eq(x0_A + v_A * t_meet, x0_B + v_B * t_meet)
t_meet_expr = sp.simplify(sp.solve(eq_meet, t_meet)[0])

# Substitute meeting time back to find meeting position x_meet relative to flagpole
x_meet_expr = sp.simplify(x0_A + v_A * t_meet_expr)

# Absolute distance from flagpole (x = 0)
d_flagpole_expr = sp.Abs(x_meet_expr)

# Render symbolic results using LaTeX
display(Math(r"\text{Meeting time: } t_{\text{meet}} = " + sp.latex(t_meet_expr)))
display(
    Math(
        r"\text{Meeting position from flagpole: } x_{\text{meet}} = "
        + sp.latex(x_meet_expr)
    )
)
display(
    Math(
        r"\text{Distance from flagpole: } d_{\text{flagpole}} = "
        + sp.latex(d_flagpole_expr)
    )
)
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical functions
calc_t_meet = sp.lambdify((x0_A, x0_B, v_A, v_B), t_meet_expr, modules="numpy")
calc_x_meet = sp.lambdify((x0_A, x0_B, v_A, v_B), x_meet_expr, modules="numpy")
calc_d_flagpole = sp.lambdify((x0_A, x0_B, v_A, v_B), d_flagpole_expr, modules="numpy")

# Instantiate quantities with AstroPy units
# Setting origin (0 mi) at flagpole: East = (+), West = (-)
x0_A_qty = -4.0 * u.imperial.mile  # 4.0 mi west of flagpole
v_A_qty = 6.0 * (u.imperial.mile / u.h)  # 6.0 mi/h due east (+)

x0_B_qty = 3.0 * u.imperial.mile  # 3.0 mi east of flagpole
v_B_qty = -5.0 * (u.imperial.mile / u.h)  # 5.0 mi/h due west (-)

# Evaluate results directly passing Quantity objects
t_meet_qty = calc_t_meet(x0_A_qty, x0_B_qty, v_A_qty, v_B_qty)
x_meet_qty = calc_x_meet(x0_A_qty, x0_B_qty, v_A_qty, v_B_qty)
d_flagpole_qty = calc_d_flagpole(x0_A_qty, x0_B_qty, v_A_qty, v_B_qty)

# Print Results
print("--- Final Results ---")
print(f"Runners meet after              : {t_meet_qty.to(u.min):.2f}")
print(f"Meeting position (from flagpole): {x_meet_qty:.2f} (negative means West)")
print(f"Distance from flagpole          : {d_flagpole_qty:.2f}")
```
