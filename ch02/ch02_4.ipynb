### Example 2.9 Not a Bad Throw for a Rookie!

page 51

**Goal** Apply the kinematic equations to a
freely falling object with a nonzero initial
velocity.

**Problem** A man throws a ball from the
top of a building with an initial velocity
of 20.0 m/s straight upward, at an initial
height of 50.0 m above the ground. The ball
just misses the edge of the roof on its way
down, as shown in Figure 2.20. Determine

* **(a)** the time needed for the ball to reach its
maximum height, 

* **(b)** the maximum height,

* **(c)** the time needed for the ball to return to
the height from which it was thrown and the
velocity of the ball at that instant, 

* **(d)** the time
needed for the ball to reach the ground, and

* **(e)** the velocity and position of the ball at
t = 5.00 s. Neglect air drag.


**Strategy** The diagram in Figure 2.20 establishes
a coordinate system with $y_0 = 0 $ at the
level at which the ball is released from the
thrower’s hand, with y positive upward. Write
the velocity and position kinematic equations
for the ball, and substitute the given information.
All the answers come from these two
equations by using simple algebra or by just
substituting the time. In part (a), for example,
the ball comes to rest for an instant at its
maximum height, so set v = 0 at this point
and solve for time. Then substitute the time
into the displacement equation, obtaining
the maximum height.

<img src="..//pics/Fig_2_20.png" width=500> 




```python
# --- Part (a) ---
v0_qty = 20.0 * (u.m / u.s)
y0_qty = 0 * u.m
h_qty = 50.0 * u.m
a_qty = -const.g0

# kinematic equation
# v = v0 + a t
# Using SymPy to solve for t
v, v0, a, y, y0 = sp.symbols("v, v_0, a,  y, y_0", real=True)
t = sp.symbols("t", real=True, positive=True)

eq_vel = sp.Eq(v, v0 + a * t)
expr_t = sp.solve(eq_vel, t)[0]
print(f"t = {expr_t}")

# at apex v=0
v_qty = 0
t_apex_qty = (v_qty - v0_qty) / a_qty

print("\n--- Part (a) ---")
print(f"Ball reaches its maximum height at time: {t_apex_qty:.3g}")
```

```python
# --- Part (b) ---
# y =  V0 t + (1/2)a t^2
y_apex_qty = v0_qty * t_apex_qty + (1 / 2) * a_qty * t_apex_qty**2

print("\n--- Part (b) ---")
print(f"Ball reaches maximum height: {y_apex_qty:.3g}")
```

```python
# --- Part (c) ---
# y =  V0 t + (1/2)a t^2

# Using SymPy to solve for t
# When ball returns to point of thrown, y=0
eq_pos = sp.Eq(0, v0 * t + (1 / 2) * a * t**2)
expr_t_c = sp.solve(eq_pos, t)[0]

print("\n--- Part (c) ---")
print(f"t = {expr_t_c}")
```

```python
print(f"Ball returns to point of throwing at time: {t_c_qty:.3g}")
# kinematic equation
# v = v0 + a t
v_c_qty = v0_qty + a_qty * t_c_qty

print(f"Ball returns to point of throwing at velocity: {v_c_qty:.3g}")
```

```python
# --- Part (d) ---
# y =  V0 t + (1/2)a t^2

# Using SymPy to solve for t
# When ball reaches ground, y=-50
eq_pos = sp.Eq(-50, v0 * t + (1 / 2) * a * t**2)
expr_t_d = sp.solve(eq_pos, t)[0]

print("\n--- Part (d) ---")
print(f"t = {expr_t_d}")
```

```python
t_ground = (
    (-v0_qty.value - np.sqrt(-100.0 * a_qty.value + v0_qty.value**2)) / a_qty.value
) * u.s

print(f"Ball reaches the ground at time: {t_ground:.3g}")
```

```python
# --- Part (e) ---
# kinematic equations
# v = v0 + a t
# y =  V0 t + (1/2)a t^2
t_e_qty = 5 * u.s

v_e_qty = v0_qty + a_qty * t_e_qty
y_e_qty = v0_qty * t_e_qty + (1 / 2) * a_qty * t_e_qty**2

print(
    f"At time {t_e_qty} the ball has velocity: {v_e_qty:.3g} and his position y: {y_e_qty:.3g}"
)
```

#### Exercise 2.9 

A projectile is launched straight up at 60.0 m/s from a height of 80.0 m, at the edge of a sheer cliff. The projectile
falls, just missing the cliff and hitting the ground below. Find 

* **(a)** the maximum height of the projectile above the point of firing,

* **(b)** the time it takes to hit the ground at the base of the cliff, and 

* **(c)** its velocity at impact.

```python
# --- Part (a) ---
# kinematic equation
# v^2 = v0 + ^2 a y
# Using SymPy to solve for y
v, v0, a, y, y0 = sp.symbols("v, v_0, a,  y, y_0", real=True)
t = sp.symbols("t", real=True, positive=True)

eq_pos = sp.Eq(v**2, v0**2 + 2 * a * y)
expr_y = sp.solve(eq_pos, y)[0]
print(f"y = {expr_y}")
```

```python
# --- Part (a) ---
v0_qty = 60.0 * (u.m / u.s)
y0_qty = 0 * u.m
h_qty = 80.0 * u.m
a_qty = -const.g0

# At apex v=0
v_apex_qty = 0 * (u.m / u.s)
y_apex_qty = (v_apex_qty**2 - v0_qty**2) / (2 * a_qty)


print("\n--- Part (a) ---")
print(
    f"The maximum height of the projectile above the point of firing y: {y_apex_qty:.0f}"
)
```

```python
# --- Part (b) ---
# y =  V0 t + (1/2)a t^2

# Using SymPy to solve for t
# When ball reaches ground, y=-80
eq_pos = sp.Eq(-80, v0 * t + (1 / 2) * a * t**2)
expr_t_b = sp.solve(eq_pos, t)[0]

print("\n--- Part (b) ---")
print(f"t = {expr_t_b}")
```

```python
t_ground_qty = (
    (-v0_qty.value - np.sqrt(-160.0 * a_qty.value + v0_qty.value**2)) / a_qty.value
) * u.s

print("\n--- Part (b) ---")
print(
    f"The time it takes to hit the ground at the base of the cliff t: {t_ground_qty:.1f}"
)
```

```python
# --- Part (c) ---
# v =  V0 + a t
v_impact_qty = v0_qty + a_qty * t_ground_qty
v_impact_book_qty = v0_qty - (9.8 * (u.m / u.s**2)) * (13.5 * u.s)
print(f"Impact velocity: {v_impact_qty:.3g}")
print(f"Impact velocity (book): {v_impact_book_qty:.3g}")
```

### Example 2.10 Maximum Height Derived

**Goal** Find the maximum height of a thrown projectile using symbols.

**Problem** Refer to Example 2.9. Use symbolic manipulation to find 

* **(a)** the time $t_{max}$ it takes the ball to reach its maximum height
and 

* **(b)** an expression for the maximum height that doesn’t depend on time. Answers should be expressed in terms of the quantities
$v_0, g$, and $y_0$ only.

**Strategy** When the ball reaches its maximum height, its velocity is zero, so for part 

* **(a)** solve the kinematics velocity equation for
time t and set $v=0$. For part 

* **(b)**, substitute the expression for time found in part (a) into the displacement equation, solving it
for the maximum height.

```python
print("\n--- Part (a) ---")


x, x0, v, v0, a, g = sp.symbols("x, x_0, v, v_0, a, g", real=True)
t, t_max = sp.symbols("t, t_max", real=True, positive=True)

# --- Velocity eq ---
eq_vel = sp.Eq(v, v0 + a * t)
print(f"\n--- Velocity equation: ---")
display(Math(rf"{sp.latex(eq_vel)}"))

print("\n --- Solve symbolically for t ---")
# At apex v=0
eq_vel_apex = sp.Eq(0, v0 + a * t)

t_apex_expr = sp.solve(eq_vel_apex, t)[0]
display(Math(rf"t = {sp.latex(t_apex_expr)}"))
# replace t = t_max, a=-g
t_substituted = t_apex_expr.subs({t: t_max, a: -g})
display(Math(rf"t_{{\text{{max}}}} = {sp.latex(t_substituted)}"))
```

```python
print("\n--- Part (a) ---")
```

```python
print("\n--- Part (b) ---")

# 1. Define the necessary symbols (making sure to include 'g' and 'y0')
y, y0, v0, a, g = sp.symbols("y y0 v0 a g", real=True)
t = sp.symbols("t", real=True, positive=True)

# 2. General position equation for y(t)
eq_pos = sp.Eq(y, y0 + v0 * t + sp.Rational(1, 2) * a * t**2)

# 3. The time expression at the apex that you calculated earlier (t = -v0 / a)
t_apex_expr = -v0 / a

# 4. Substitute t with the apex time expression into the right-hand side of the position equation
y_max_expr = eq_pos.rhs.subs(t, t_apex_expr)

print("--- Substituting t ---")
display(Math(rf"y_{{\text{{max}}}} = {sp.latex(y_max_expr)}"))

# 5. Simplify the expression algebraically (combines the like terms, similar to the textbook's middle step)
y_max_simplified = sp.simplify(y_max_expr)

# 6. Substitute a = -g as indicated in the textbook
y_max_final = y_max_simplified.subs(a, -g)

print("--- Substituting a = -g and simplifying ---")
display(Math(rf"y_{{\text{{max}}}} = {sp.latex(y_max_final)}"))
```

### Example 2.11 A Rocket Goes Ballistic

**Goal** Solve a problem involving a powered ascent followed by free-fall motion.

**Problem** A rocket moves straight upward, starting from rest with an
acceleration of $+29.4 m/s^2$. It runs out of fuel at the end of 4.00 s and
continues to coast upward, reaching a maximum height before falling back
to Earth. 

* **(a)** Find the rocket’s velocity and position at the end of 4.00 s.

* (**b)** Find the maximum height the rocket reaches. 

* **(c)** Find the velocity the instant before the rocket crashes
on the ground.

**Strategy** Take y = 0 at the launch point and y positive upward, as in
Figure 2.21. The problem consists of two phases. In phase 1 the rocket
has a net upward acceleration of $29.4 m/s^2$, and we can use the kinematic
equations with constant acceleration a to find the height and
velocity of the rocket at the end of phase 1, when the fuel is burned up. 
In phase 2 the rocket is in free fall and has an acceleration of $-9.80 m/s^2$, with initial velocity
and position given by the results of phase 1. Apply the kinematic equations for free fall.

<img src="..//pics/Fig_2_21.png" width=700> 


```python
# --- Phase 1 ---
# --- rocket has a constant acceleration ---
a1_qty = 29.4 * (u.m / u.s**2)
t1_qty = 4.00 * u.s

# vel eq: v = v0 + at
v1_qty = 0 + a1_qty * t1_qty

print("--- Part (a) ---")
print(f"At time {t1_qty:.3g} the rocket has a velocity of v1 = {v1_qty:.3g}")


# pos eq: y = y0 + vo t + (1/2) a t^2
y1_qty = 0 + 0 + (1 / 2) * a1_qty * t1_qty**2
print(f"At time {t1_qty:.3g} the rocket has a position y1 = {y1_qty:.3g}")
```

```python
# --- Phase 2 ---
# --- rocket is in free fall ---

a, g, v, v0, x, x0, y, y0 = sp.symbols("a, g, v, v0, x, x0, y, y0", real=True)

# --- Velocity square eq ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * y)
print(f"\n--- Velocity square equation: ---")
display(Math(rf"{sp.latex(eq_vel_sq)}"))

print(" --- Solve symbolically for y ---")
y_vel_expr = sp.solve(eq_vel_sq, y)[0]
display(Math(rf"y = {sp.latex(y_vel_expr)}"))
```

```python
# at apex v=0, x=y
v2_qty = 0 * (u.m / u.s)
v02_qty = v1_qty
a2_qty = -const.g0
# from the symbolic solution
y2_qty = (v2_qty**2 - v02_qty**2) / (2 * a2_qty)


print("--- Part (b) ---")
print(f"At phase two the rocket has reach y2 = {y2_qty:.3g} more")

y_max_qty = y1_qty + y2_qty
print(f"At apex the rocket has reach y_max = {y_max_qty:.3g}")

# NOTE: there is a 5 m difference with the book, as I don't use rounded intermediate results.
```

```python
a, g, v, v0, x, x0, y, y0 = sp.symbols("a, g, v, v0, x, x0, y, y0", real=True)

# --- Velocity square eq ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * y)
print(f"\n--- Velocity square equation: ---")
display(Math(rf"{sp.latex(eq_vel_sq)}"))

print(" --- Solve symbolically for v ---")
v_vel_expr = sp.solve(eq_vel_sq, v)[0]
display(Math(rf"v = {sp.latex(v_vel_expr)}"))
```

```python
# The book solve this part with two equations
# We solve it with one equation, taken into account
# a free fall from apex to the floor

# v0 = 0
v_crash_qty = -np.sqrt(2 * const.g0 * y_max_qty + 0)

print("--- Part (c) ---")
print(f"The rocket crashes at the floor with velocity v_crash = {v_crash_qty:.3g}")
```

#### Exercise 2.11 
An experimental rocket designed to land upright falls freely from a height of $2.00 \times 10^2 m$, starting at rest. At a
height of 80.0 m, the rocket’s engines start and provide constant upward acceleration until the rocket lands. What acceleration is
required if the speed on touchdown is to be zero? (Neglect air resistance.)

```python
# --- Phase 1 ---
# --- rocket is in free fall until y1 = 80 m. ---
# --- Calculate velocity at 80 m. (we use the last formula calculated)
print("--- Phase 1 ---")

display(Math(rf"v = {sp.latex(v_vel_expr)}"))
y_max_qty = 2.00e2 * u.m  # the apex
y_se_qty = 80 * u.m  # starting engines
y_ph1_qty = y_max_qty - y_se_qty  # free fall in phase 1

v_80_qty = -np.sqrt(2 * const.g0 * y_ph1_qty + 0)

print(
    f"When the rocket start engines at {y_se_qty} its velocity is: v_80 = {v_80_qty:.3g}"
)
```

```python
# --- Phase 2 ---
# --- At 80 m, rocket starts engines ---
# --- Calculate acceleration to touch down with v=0
print("--- Phase 1 ---")

a, g, v, v0, x, x0, y, y0 = sp.symbols("a, g, v, v0, x, x0, y, y0", real=True)

# --- Velocity square eq ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * y)
print(f"\n--- Velocity square equation: ---")
display(Math(rf"{sp.latex(eq_vel_sq)}"))

print(" --- Solve symbolically for a ---")
v_a_expr = sp.solve(eq_vel_sq, a)[0]
display(Math(rf"a = {sp.latex(v_a_expr)}"))

a_qty = (0 - v_80_qty**2) / (2 * y_se_qty)
print(f"The rocket required acceleration: a = {a_qty:.3g}")
```
