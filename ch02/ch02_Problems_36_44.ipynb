#### 36.

A record of travel along a straight path is as follows:

1. Start from rest with a constant acceleration of 2.77 m/s2
for 15.0 s.

2. Maintain a constant velocity for the next 2.05 min.

3. Apply a constant negative acceleration of 29.47 m/s2
for 4.39 s.

* **(a)** What was the total displacement for the trip?

* **(b)** What were the average speeds for legs 1, 2, and 3 of the
trip, as well as for the complete trip?


```python
# --- Leg 1 ---
a1_qty = 2.77 * (u.m / u.s**2)
t1_qty = 15.0 * u.s

# --- Leg 2 ---
t2_qty = (2.05 * u.min).to(u.s)

# --- Leg 3 ---
a3_qty = -29.47 * (u.m / u.s**2)
t3_qty = 4.39 * u.s

print("--- Part (a)---")
# --- calculate displacement for each leg; eq: dx = x0 + v0 t + (1/2) a t^2 ---
d1_qty = (1 / 2) * a1_qty * t1_qty**2
print("\nLeg 1:")
print(f"Displacement in Leg 1: d1 = {d1_qty:.2f}")

# eq: v = v0 + a t
v2_qty = 0 + a1_qty * t1_qty
print("\nLeg 2:")
print(f"The constant velocity in this leg is v={v2_qty:.2f}")
d2_qty = v2_qty * t2_qty + 0
print(f"Displacement in Leg 2: d2 = {d2_qty:.2f}")

# eq: x = v0 * t + (1/2) a t^2
d3_qty = (v2_qty * t3_qty) + ((1 / 2) * a3_qty * t3_qty**2)
print(f"Displacement in Leg 3: d3 = {d3_qty:.2f}")


# total displacement = d = d1 + d2 + d3
d_qty = d1_qty + d2_qty + d3_qty
print(f"The total displacement: d = {d_qty:.2f}")
```

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic parameters
a1, t1 = sp.symbols("a1 t1", real=True, positive=True)
t2 = sp.symbols("t2", real=True, positive=True)
a3, t3 = sp.symbols("a3 t3", real=True, positive=True)

# Leg 1
x1 = sp.Rational(1, 2) * a1 * t1**2
v1 = a1 * t1

# Leg 2
x2 = v1 * t2

# Leg 3
x3 = v1 * t3 + sp.Rational(1, 2) * a3 * t3**2

# Total Displacement
x_total = sp.simplify(x1 + x2 + x3)

# Total Time
t_total = t1 + t2 + t3

# Average speeds
v_avg1 = sp.simplify(x1 / t1)
v_avg2 = v1
v_avg3 = sp.simplify(x3 / t3)
v_avg_total = sp.simplify(x_total / t_total)

# Render symbolic results using LaTeX
display(
    Math(r"\text{Total Displacement: } \Delta x_{\text{total}} = " + sp.latex(x_total))
)
display(
    Math(
        r"\text{Complete Trip Average Speed: } v_{\text{avg}} = "
        + sp.latex(v_avg_total)
    )
)
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical functions
calc_x_total = sp.lambdify((a1, t1, t2, a3, t3), x_total, modules="numpy")
calc_v_avg1 = sp.lambdify((a1, t1), v_avg1, modules="numpy")
calc_v_avg2 = sp.lambdify((a1, t1), v_avg2, modules="numpy")
calc_v_avg3 = sp.lambdify((a1, t1, a3, t3), v_avg3, modules="numpy")
calc_v_avg_total = sp.lambdify((a1, t1, t2, a3, t3), v_avg_total, modules="numpy")

# Instantiate quantities with AstroPy units
a1_qty = 2.77 * (u.m / (u.s**2))
t1_qty = 15.0 * u.s
t2_qty = 2.05 * u.min
a3_qty = -9.47 * (u.m / (u.s**2))
t3_qty = 4.39 * u.s

# Evaluate results directly passing Quantity objects
x_tot_qty = calc_x_total(a1_qty, t1_qty, t2_qty, a3_qty, t3_qty)
v_a1_qty = calc_v_avg1(a1_qty, t1_qty)
v_a2_qty = calc_v_avg2(a1_qty, t1_qty)
v_a3_qty = calc_v_avg3(a1_qty, t1_qty, a3_qty, t3_qty)
v_a_tot_qty = calc_v_avg_total(a1_qty, t1_qty, t2_qty, a3_qty, t3_qty)

# Print Results
print("--- Final Results ---")
print(f"(a) Total displacement       : {x_tot_qty.to(u.m):.2f}")
print(f"(b) Average speed for Leg 1  : {v_a1_qty.to(u.m / u.s):.2f}")
print(f"    Average speed for Leg 2  : {v_a2_qty.to(u.m / u.s):.2f}")
print(f"    Average speed for Leg 3  : {v_a3_qty.to(u.m / u.s):.2f}")
print(f"    Complete trip avg speed  : {v_a_tot_qty.to(u.m / u.s):.2f}")
```

#### 37. 
A train is traveling down a straight track at 20 m/s when the
engineer applies the brakes, resulting in an acceleration of
-1.0 m/s2 as long as the train is in motion. How far does the
train move during a 40-second time interval starting at the
instant the brakes are applied?

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
v_0, a, t_interval = sp.symbols("v_0 a t_interval", real=True, positive=True)

# Time to stop: v_0 + a * t_stop = 0  =>  t_stop = -v_0 / a (using magnitude of a)
# Since a is deceleration, let a_mag be the magnitude
a_mag = sp.symbols("a_mag", real=True, positive=True)

t_stop = v_0 / a_mag

# Displacement until stop: delta_x = v_0 * t_stop - (1/2) * a_mag * t_stop^2
delta_x_expr = sp.simplify(v_0 * t_stop - sp.Rational(1, 2) * a_mag * t_stop**2)

# Render symbolic results using LaTeX
display(Math(r"\text{Time to stop: } t_{\text{stop}} = " + sp.latex(t_stop)))
display(
    Math(
        r"\text{Total displacement during interval: } \Delta x = "
        + sp.latex(delta_x_expr)
    )
)
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical function
calc_delta_x = sp.lambdify((v_0, a_mag), delta_x_expr, modules="numpy")

# Instantiate quantities with AstroPy units
v_0_qty = 20.0 * (u.m / u.s)
a_mag_qty = 1.0 * (u.m / (u.s**2))
t_interval_qty = 40.0 * u.s

# Evaluate numerical results directly passing Quantity objects
delta_x_qty = calc_delta_x(v_0_qty, a_mag_qty)

# Also check stopping time condition
t_stop_qty = v_0_qty / a_mag_qty

# Print Results
print("--- Final Results ---")
print(f"Time to come to rest : {t_stop_qty.to(u.s):.1f}")
if t_interval_qty >= t_stop_qty:
    print(f"Total distance moved (in 40 s) : {delta_x_qty.to(u.m):.1f}")
else:
    # If interval was shorter than stopping time
    pass
```

#### 38. 

A car accelerates uniformly from rest to a speed of 40.0 mi/h
in 12.0 s. Find 

* **(a)** the distance the car travels during this time
and 

* **(b)** the constant acceleration of the car.

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
v_i, v_f, t = sp.symbols("v_i v_f t", real=True, positive=True)
a, delta_x = sp.symbols("a delta_x", real=True, positive=True)

# (b) Symbolic acceleration: v_f = v_i + a * t  =>  a = (v_f - v_i) / t
eq_a = sp.Eq(v_f, v_i + a * t)
a_expr = sp.simplify(sp.solve(eq_a, a)[0])

# (a) Symbolic distance: delta_x = v_i * t + (1/2) * a * t^2
eq_x = sp.Eq(delta_x, v_i * t + sp.Rational(1, 2) * a * t**2)
delta_x_expr = sp.simplify(eq_x.rhs.subs(a, a_expr))

# Render symbolic results using LaTeX
display(Math(r"\text{(a) Symbolic distance: } \Delta x = " + sp.latex(delta_x_expr)))
display(Math(r"\text{(b) Symbolic acceleration: } a = " + sp.latex(a_expr)))
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical functions
calc_delta_x = sp.lambdify((v_i, v_f, t), delta_x_expr, modules="numpy")
calc_a = sp.lambdify((v_i, v_f, t), a_expr, modules="numpy")

# Instantiate quantities with AstroPy units
v_i_qty = 0.0 * (u.m / u.s)
v_f_qty = 40.0 * (u.imperial.mi / u.h)
t_qty = 12.0 * u.s

# Evaluate numerical results directly passing Quantity objects
delta_x_qty = calc_delta_x(v_i_qty, v_f_qty, t_qty)
a_qty = calc_a(v_i_qty, v_f_qty, t_qty)

# Print Results
print("--- Final Results ---")
print(f"(a) Distance traveled : {delta_x_qty.to(u.m):.1f}")
print(f"(b) Constant acceleration : {a_qty.to(u.m / (u.s**2)):.2f}")
```

#### 39. 
A car starts from rest and travels for 5.0 s with a uniform acceleration
of 11.5 m/s2. The driver then applies the brakes, causing
a uniform acceleration of -2.0 m/s2. If the brakes are
applied for 3.0 s, 

* **(a)** how fast is the car going at the end of the
braking period, and 

* **(b)** how far has the car gone?

```python
t1_qty = 5.0 * u.s
a1_qty = +1.5 * (u.m / u.s**2)
a2_qty = -2.0 * (u.m / u.s**2)
t2_qty = 3.0 * u.s

# --- Part (a) ---
# eq: v = v0 + a t
v1_qty = 0 + a1_qty * t1_qty

v2_qty = v1_qty + (a2_qty * t2_qty)

# --- Part (b) ---
# eq: x = v0 t + (1/2) a t^2
l1_qty = (1 / 2) * a1_qty * t1_qty**2  # leg 1 (acceleration period)
l2_qty = (v1_qty * t2_qty) + ((1 / 2) * a2_qty * t2_qty**2)  # leg 2 (breaking period)
delta_x_qty = l1_qty + l2_qty


# --- results ---
print("\n--- Part (a) ---")
print(f"The car, at the end of braking period, has a velocity: {v2_qty}")

print("\n--- Part (b) ---")
print(f"The car has travelled: {delta_x_qty}")
```

#### 41. 
In a car race, a Ford Thunderbird and a Mercedes Benz are
moving side by side down a straight track at 71.5 m/s. The
Thunderbird driver makes a pit stop, slowing her car to rest
at uniform acceleration over a distance of 250 m, while the
Mercedes continues on at constant speed. After 5.00 s at rest,
she accelerates uniformly to her previous speed of 71.5 m/s,
this time covering a distance of 350 m. How far has she fallen
behind the Mercedes Benz?

```python
x, x0, v, v0, a = sp.symbols("x, x_0, v, v_0, a", real=True)
t = sp.symbols("t", real=True, positive=True)
# --- Velocity square eq ---
eq_vel_sq = sp.Eq(v**2, v0**2 + 2 * a * x)
print(f"\n--- Velocity square equation: ---")
display(Math(rf"{sp.latex(eq_vel_sq)}"))

print("\n --- Solve symbolically for a ---")
a_vel_sq_expr = sp.solve(eq_vel_sq, a)[0]
display(Math(rf"a = {sp.latex(a_vel_sq_expr)}"))

# --- Velocity eq ---
eq_vel = sp.Eq(v, v0 + a * t)
print(f"\n--- Velocity equation: ---")
display(Math(rf"{sp.latex(eq_vel)}"))

print("\n --- Solve symbolically for t ---")
t_vel_expr = sp.solve(eq_vel, t)[0]
display(Math(rf"t = {sp.latex(t_vel_expr)}"))
```

```python
# Derive callable numerical functions
calc_a = sp.lambdify((v, v0, x), a_vel_sq_expr, modules="numpy")
calc_t = sp.lambdify((v, v0, a), t_vel_expr, modules="numpy")

# AstroPy units
v0_FT_qty = 71.5 * (u.m / u.s)
v0_MB_qty = 71.5 * (u.m / u.s)
x_l1_qty = 250 * u.m
t_break = 5.00 * u.s
x_l2_qty = 350 * u.m

# Ford Thunderbird legs, calculate the time
# --- L1 breaking to stop during 250 m, v=0 ---
v1_qty = 0.0 * (u.m / u.s)

a1_qty = calc_a(v1_qty, v0_FT_qty, x_l1_qty)
t1_qty = calc_t(v1_qty, v0_FT_qty, a1_qty)

# --- L2 acceleration during 350 m, v=71.5 m/s ---
v02_qty = 0.0 * (u.m / u.s)

a2_qty = calc_a(v0_FT_qty, v02_qty, x_l2_qty)
t2_qty = calc_t(v0_FT_qty, v02_qty, a2_qty)

t_total_qty = t1_qty + t_break + t2_qty

# calculate the distance that TB car has covered
x_TB_qty = x_l1_qty + x_l2_qty


# calculate the distance that MB car has covered
# eq: v = x / t ==> x = v t
x_MB_qty = v0_MB_qty * t_total

x_behind_qty = x_MB_qty - x_TB_qty


# --- results ---
print(f"The Ford Thunderbird acceleration in leg 1: {a1_qty:.3g}")
print(f"The Ford Thunderbird time in leg 1: {t1_qty:.3g}")

print(f"The Ford Thunderbird acceleration in leg 2: {a2_qty:.3g}")
print(f"The Ford Thunderbird time in leg 2: {t2_qty:.3g}")
print(
    f"The Ford Thunderbird total time in the legs plus the breaking: {t_total_qty:.3g}"
)
print(f"The Ford Thunderbird total distance in total time: {x_TB_qty:.3g}")
print(f"The Mercedes Benz total distance in total time: {x_MB_qty:.0f}")
print()
print(f"Final answer: She has fallen behind: {x_behind_qty:.3g}")
```

#### 42
The kinematic equations can describe phenomena other than
motion through space and time. Suppose x represents a person’s
bank account balance. The units of x would be dollars
(+), and velocity v would give the rate at which the balance
changes (in units of, for example, +/month). Acceleration
would give the rate at which v changes. Suppose a person
begins with ten thousand dollars in the bank. Initial money
management leads to no net change in the account balance
so that $v_0 = 0$. Unfortunately, management worsens over time
so that  $a =  -2.5 \times 10^2 \$/month^2$. Assuming a is constant,
find the amount of time in months until the bank account is
empty.

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
x_0, v_0, a, t = sp.symbols("x_0 v_0 a t", real=True, positive=True)

# Position equation: 0 = x_0 + v_0 * t - (1/2) * a_mag * t^2
# (treating 'a' as the positive magnitude of negative acceleration)
eq_pos = sp.Eq(0, x_0 + v_0 * t - sp.Rational(1, 2) * a * t**2)

# Solve symbolically for time t
t_solutions = sp.solve(eq_pos, t)
t_expr = t_solutions[0]  # Positive physical root

# Render symbolic result using LaTeX
display(Math(r"\text{Symbolic time until account is empty: } t = " + sp.latex(t_expr)))
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical function
calc_t = sp.lambdify((x_0, v_0, a), t_expr, modules="numpy")

# Define custom monetary/time units for AstroPy
dollar = u.def_unit("dollar")
month = u.def_unit("month")

# Instantiate quantities with AstroPy units
x_0_qty = 10000.0 * dollar
v_0_qty = 0.0 * (dollar / month)
a_qty = 250.0 * (dollar / (month**2))  # Magnitude of worsening acceleration

# Evaluate results directly passing Quantity objects
t_qty = calc_t(x_0_qty, v_0_qty, a_qty)

# Print Results
print("--- Final Results ---")
print(f"Time until bank account is empty: {t_qty:.2f}")
```

#### 43. 

A hockey player is standing on his skates on a frozen
pond when an opposing player, moving at a constant speed of
8.0 m/s, skates by with the puck. After 0.25 s, the first player
gives chase, accelerating uniformly at 4.0 m/s2. 

* **(a)** How long
does it take him to catch his opponent? 

* **(b)** How far has he
traveled in that time?

```python
x, x0, v, v0, a = sp.symbols("x, x_0, v, v_0, a", real=True)
t = sp.symbols("t", real=True, positive=True)
# --- Position eq ---
eq_pos = sp.Eq(x, x0 + v0 * t + sp.Rational(1, 2) * a * t**2)
print(f"\n--- Position equation: ---")
display(Math(rf"{sp.latex(eq_pos)}"))

# eq position P1 ==> x = x0 + v0 t
x1_expr = x0 + v0 * t
# eq position P2 ==> x = (1/2) a t^2
x2_expr = sp.Rational(1, 2) * a * t**2
# player 2 chase player 1 when x1 = x2
print(f"\n--- Chase equation: ---")
chase_eq = sp.Eq(x1_expr, x2_expr)
display(Math(rf"{sp.latex(chase_eq)}"))

print("\n --- Solve symbolically for t ---")
t_pos_expr = sp.solve(chase_eq, t)[1]
display(Math(rf"t = {sp.latex(t_pos_expr)}"))
```

```python
v0_P1_qty = 8.0 * (u.m / u.s)  # velocity Player 1 - P1
t_reaction_qty = 0.25 * u.s  # time reaction Player 2 - P2
a_P2_qty = 4.0 * (u.m / u.s**2)  # acceleration Player 2 - P2

# --- Part (a) ---

# --- Calculate the distance covered by Player 1, until Player 2 reacts
# eq: v = x / t ==> x = v t
x0_P1_qty = v0_P1_qty * t_reaction_qty

# Derive callable numerical function
calc_t = sp.lambdify((v0, a, x0), t_pos_expr, modules="numpy")

t_chase_qty = calc_t(v0_P1_qty, a_P2_qty, x0_P1_qty)


# --- Part (b) ---
# eq: x = x0 + v0 t + (1/2) a t^2 (x0=0, v0=0) ==> x = (1/2) a t^2
x_P2_qty = (1 / 2) * a_P2_qty * t_chase_qty**2


# --- results ---

print("\n--- Part (a) ---")
print(f"Player 2 catches Player 1 in time: {t_chase_qty:.2g}")

print("\n--- Part (b) ---")
print(f"Player 2 has covered a distance of: {x_P2_qty:.3g}")
```

#### 44. 

A train $4.00 \times 10^2 m$ long is moving on a straight track with
a speed of 82.4 km/h. The engineer applies the brakes at a
crossing, and later the last car passes the crossing with a speed
of 16.4 km/h. Assuming constant acceleration, determine
how long the train blocked the crossing. Disregard the width
of the crossing.

```python
# --------------------------------------------------
# Stage 1: Symbolic Derivation (SymPy)
# --------------------------------------------------

# Define symbolic variables
v0, vf, L, a, t = sp.symbols("v0 vf L a t", real=True, positive=True)

# 1. Kinematic equation for acceleration: vf^2 = v0^2 + 2*a*L
eq_a = sp.Eq(vf**2, v0**2 + 2 * a * L)
a_expr = sp.simplify(sp.solve(eq_a, a)[0])

# 2. Kinematic equation for blocking time: L = (1/2)*(v0 + vf)*t
eq_t = sp.Eq(L, sp.Rational(1, 2) * (v0 + vf) * t)
t_expr = sp.simplify(sp.solve(eq_t, t)[0])

# Render symbolic results using LaTeX
display(Math(r"\text{Symbolic acceleration: } a = " + sp.latex(a_expr)))
display(Math(r"\text{Symbolic blocking time: } t = " + sp.latex(t_expr)))
```

```python
# --------------------------------------------------
# Stage 2: Numerical Evaluation (AstroPy + lambdify)
# --------------------------------------------------

# Derive callable numerical functions
calc_a = sp.lambdify((vf, v0, L), a_expr, modules="numpy")
calc_t = sp.lambdify((vf, v0, L), t_expr, modules="numpy")

# Instantiate quantities with AstroPy units
L_qty = 4.00e2 * u.m  # Train length: 400 m
v0_qty = 82.4 * (u.km / u.h)  # Initial speed as front reaches crossing
vf_qty = 16.4 * (u.km / u.h)  # Final speed as last car leaves crossing

# Evaluate results directly passing Quantity objects
a_qty = calc_a(vf_qty, v0_qty, L_qty)
t_qty = calc_t(vf_qty, v0_qty, L_qty)

# Print Results
print("--- Final Results ---")
print(f"Train acceleration         : {a_qty.to(u.m / (u.s**2)):.3f}")
print(f"Time train blocked crossing : {t_qty.to(u.s):.2f}")
```
