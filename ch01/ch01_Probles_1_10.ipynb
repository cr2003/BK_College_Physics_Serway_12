The Roman cubitus is an ancient unit of measure equivalent
to about 0.445 m. Convert the 2.00-m height of a basketball
forward to cubiti.


```python
# 1. Define the custom unit: 1 Roman cubitus = 0.445 meters
cubitus = u.def_unit("cubiti", 0.445 * u.m)

# 2. Set the basketball player's height
height_qty = 2.00 * u.m

# 3. Convert to cubiti
height_in_cubiti = height_qty.to(cubitus)

# 4. Display formatted result with significant figures (3 sig figs)
height_in_cubiti
print(f"Height: {height_in_cubiti:.3g}")
```

#### 8. 
A house is advertised as having 1 420 square feet under roof.
What is the area of this house in square meters?

```python
A_ft_qty = 1_420 * (u.imperial.ft**2)
A_m2_qty = A_ft_qty.to(u.m**2)

print(f"Area: {A_m2_qty:.3g}")
```

#### 9. 
A fathom is a unit of length, usually reserved for measuring
the depth of water. A fathom is approximately 6 ft in
length. Take the distance from Earth to the Moon to be
250 000 miles, and use the given approximation to find the
distance in fathoms.

```python
# 1. Define the custom unit: 1 Roman cubitus = 0.445 meters
fathom = u.def_unit("fathom", 6 * u.imperial.ft)

dem_mi_qty = 250_000 * u.imperial.mi  # dem - distance earth moon
dem_fth_qyt = dem_mi_qty.to(fathom)

print(f"Distance Earth to Moon: {dem_fth_qyt:.0e}")
```

#### 10. 
A small turtle moves at a speed of 186 furlongs per fortnight.
Find the speed of the turtle in centimeters per second. Note
that 1 furlong = 220 yards and 1 fortnight = 14 days.

```python
# 1. Define the custom unit: 1 fortnight = 14 days
fn = u.def_unit("fortnight", 14 * u.d)

speed_fnpfur_qty = 186 * (u.imperial.furlong / fn)

speed_cmps_qty = speed_fnpfur_qty.to(u.cm / u.s)

print(f"Turtle speed: {speed_cmps_qty:.2g}")
```
