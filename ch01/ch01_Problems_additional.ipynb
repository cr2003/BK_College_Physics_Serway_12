**(a)** Find a conversion factor to convert from miles per hour
to kilometers per hour. 

**(b)** For a while, federal law mandated
that the maximum highway speed would be 55 mi/h. Use the
conversion factor from part (a) to find the speed in kilometers
per hour. 

**(c)** The maximum highway speed has been raised to
65 mi/h in some places. In kilometers per hour, how much of
an increase is this over the 55-mi/h limit?


```python
# --- (a) ---
mph = 1 * (u.imperial.mi / u.h)
ratio = mph.to(u.km / u.h)

# --- (b) ---
mhs_mph = 55 * (u.imperial.mi / u.h)  # maximum highway speed in miles per hour
mhs_kph = mhs_mph.to(u.km / u.h)

# --- (c) ---
mhs_mph2 = 65 * (u.imperial.mi / u.h)  # maximum highway speed in miles per hour
mhs_kph2 = mhs_mph2.to(u.km / u.h)

increase = mhs_kph2 - mhs_kph


print(f"\n--- (a) ---")
print(f"ratio = {ratio.value:.5g}")

print(f"\n--- () ---")
print(f"{mhs_mph:.0f} = {mhs_kph:.0f}")

print(f"\n--- () ---")
print(f"Increase = {increase:.0f}")
```

#### 74. 
Assume it takes 7.00 minutes to fill a 30.0-gal gasoline
tank. 

**(a)** Calculate the rate at which the tank is filled in gallons
per second. 

**(b)** Calculate the rate at which the tank is filled
in cubic meters per second.

**(c)** Determine the time interval,
in hours, required to fill a 1.00-m3 volume at the same rate.
(1 U.S. gal 5 231 in.3)

```python
t_qty = 7 * u.min
v_qty = 30.0 * u.imperial.gallon
rate_gpm_qty = v_qty / t_qty  # rate gallons per minute

# --- (a) ---
rate_gps_qty = rate_gpm_qty.to(u.imperial.gallon / u.s)


# --- (b) ---
rate_cmps_qty = rate_gpm_qty.to(u.m**3 / u.s)

# --- (c) ---
v1_qty = 1.00 * u.m**3
rate_cbph_qty = rate_gpm_qty.to(u.m**3 / u.h)
# eq: r = v / t ==> t = v / r
t_qty = v1_qty / rate_cbph_qty


# --- results ---
print(f"\n--- (a) ---")
print(f"Rate: {rate_gps_qty:.2e}")

print(f"\n--- (b) ---")
print(f"Rate: {rate_cmps_qty:.2e}")

print(f"\n--- (c) ---")
print(f"Time: {t_qty:.3g}")
```

#### 75. 

One gallon of paint (volume =  $3.79 \times 10^{-3}\, m^3$) covers
an area of $25.0 \,m^2$. What is the thickness of the fresh paint on
the wall?

```python
bf_qty = 3.79e-3 * u.m**3 / (1 * u.imperial.gallon)  # book factor conversion
a_qty = 25.0 * u.m**2

# --- conversions ---
V_qty = 1 * u.imperial.gallon
V_m3_bf_qty = V_qty * bf_qty

# --- calculations ---
# eq: V = area * thickness ==> tickness = V / area
thickness_bf_qty = V_m3_bf_qty / a_qty

# --- results ---
print(f"thicknes (bf): {thickness_bf_qty.to(u.um):.0f}")
```

#### 76. 
A sphere of radius r has surface area $A = 4 \pi r^2$ and volume
$V = 4/3 \pi r^3$. If the radius of sphere 2 is double the radius
of sphere 1, what is the ratio of 

**(a)** the areas, A2/A1 and

**(b)** the volumes, V2 /V1 ?

```python
# Define symbolic variables
r1 = sp.Symbol("r1", real=True, positive=True)

# Radius of sphere 2 is double the radius of sphere 1
r2 = 2 * r1

# Surface area and volume formulas
A1 = 4 * sp.pi * r1**2
A2 = 4 * sp.pi * r2**2

V1 = (sp.Rational(4, 3)) * sp.pi * r1**3
V2 = (sp.Rational(4, 3)) * sp.pi * r2**3

# (a) Ratio of areas A2 / A1
ratio_A = sp.simplify(A2 / A1)

# (b) Ratio of volumes V2 / V1
ratio_V = sp.simplify(V2 / V1)

print("--- Symbolic Results ---")
print(f"A2 / A1 = {ratio_A}")
print(f"V2 / V1 = {ratio_V}")
```

#### 77. 

Assume there are 250 million passenger cars in Europe
and that the average fuel consumption is 6 L/100 km. If the
average distance traveled by each car is 15 000 km/yr, how
much fuel would be saved each year if average fuel consumption
could be increased to 4.5 L/100 km? Answer with
one significant figure.

```python
num_cars = 250e6
cons_1 = (6 * u.L) / (100 * u.km)
d_car = 15_000 * (u.km / u.yr)
cons_2 = (4.5 * u.L) / (100 * u.km)

f1 = num_cars * cons_1 * d_car
f2 = num_cars * cons_2 * d_car

saved = f1 - f2
print(f"Fuel saved {saved:.1g}")
```

#### 78. 

In 2015, the U.S. national debt was about $18 trillion.

**(a)** If payments were made at the rate of $1 000 per second,
how many years would it take to pay off the debt, assuming
that no interest were charged? 

**(b)** A dollar bill is about
15.5 cm long. If 18 trillion dollar bills were laid end to end
around the Earth’s equator, how many times would they
encircle the planet? Take the radius of the Earth at the
equator to be 6 378 km. (Note: Before doing any of these
calculations, try to guess at the answers. You may be very
surprised.)

```python
# 1. Define custom currency unit
USD = u.def_unit("USD")
usnd = 18e12 * USD  # American trillion 10^12

# --- (a) ---
payment_rate = 1_000 * (USD / u.s)
pay_off = usnd / payment_rate

# --- (b) ---
dbl = 15.5 * (u.cm / USD)  # dollar bill length
r_earth = 6_378 * u.km
circ_earth = 2 * np.pi * r_earth
rounds = ((usnd * dbl) / circ_earth).decompose()


# --- results ---
print("\n--- (a) ---")
print(f"Pay off time: {pay_off.to(u.yr):.0f}")
pay_off.to(u.yr)

print("\n--- (b) ---")
print(f"Earth encirclements: {rounds:,.0f}")
```

An average person sneezes about three times per day.
Estimate the worldwide number of sneezes happening in a
time interval approximately equal to one sneeze.

```python
# 1. Define symbolic calculation
rate_person, N_pop, t_sneeze = sp.symbols(
    "rate_person N_pop t_sneeze", real=True, positive=True
)
N_total_expr = N_pop * rate_person * t_sneeze

calc_N = sp.lambdify((rate_person, N_pop, t_sneeze), N_total_expr)

# --- Define quantities with AstroPy units ---
rate_person_qty = 3.0 / (u.day)  # 3 sneezes per person per day
N_pop_qty = 8.3e9  # World population (~8 billion) (yr. 2026)
t_sneeze_qty = 1.0 * u.s  # Duration of one sneeze

# --- Compute ---
N_qty = rate_person_qty * N_pop_qty * t_sneeze_qty
N_qty = N_qty.decompose()

# 4. Print using .value to apply float formatting string safely
print("--- AstroPy Results ---")
print(f"Total worldwide sneezes: {N_qty:.3g}")
```

The nearest neutron star (a collapsed star made primarily of
neutrons) is about $ 3.00 \times 10^{18} \;m$ away from Earth. Given that
the Milky Way galaxy (Fig. P1.81) is roughly a disk of diameter
~ $10^{21}\; m$ and thickness , ~ $10^{19}\;$,  estimate the number of neutron
stars in the Milky Way to the nearest order of magnitude.

<img src="..//pics/Fig_P1_81.png" width=300> 


```python
r_ns_qty = 3.00e18 * u.m  # radius to nearest neutron star
r_mkw_qty = 1e21 / 2 * u.m  # radius of the Milky Way
h_mkw_qty = 1e19 * u.m  # thickness of the Milky Way

v_ns_qty = (4 / 3) * np.pi * r_ns_qty**3  # Volume to find a neutron star
v_mkw_qty = np.pi * r_mkw_qty**2 * h_mkw_qty  # Volume of the Milky Way

n_ns_qty = v_mkw_qty / v_ns_qty  # number of neutron stars in the Milky Way

print(f"Calculated neutron stars in the Milky Way: {n_ns_qty:.3g}")
print(f"Orden of Magnitude: {oom(n_ns_qty):.0e}")
```
