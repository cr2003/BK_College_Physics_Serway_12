#### 44.
Given points $(r1, \theta1)$ and $(r2, \theta2)$ in polar coordinates,
obtain a general formula for the distance between them. Simplify
it as much as possible using the identity $cos^2 \theta + sin^2 \theta = 1$.
Hint: Write the expressions for the two points in Cartesian
coordinates and substitute into the usual distance formula.

```python
# Define symbolic variables
r1, r2, theta1, theta2 = sp.symbols("r1 r2 theta1 theta2", real=True, positive=True)

# Cartesian components
x1, y1 = r1 * sp.cos(theta1), r1 * sp.sin(theta1)
x2, y2 = r2 * sp.cos(theta2), r2 * sp.sin(theta2)

# Distance squared formula
d_squared = (x2 - x1) ** 2 + (y2 - y1) ** 2

# Trigonometric simplification
d_squared_simplified = sp.trigsimp(d_squared)

print("d^2 =", d_squared_simplified)
# Output: d^2 = r1**2 - 2*r1*r2*cos(theta1 - theta2) + r2**2

print("NOTE: thats the result is the Law of Cosines")
```

#### 46. 
A ladder 9.00 m long leans against the side of a building. If
the ladder is inclined at an angle of 75.0° to the horizontal,
what is the horizontal distance from the bottom of the ladder
to the building?

```python
l_qty = 9.00 * u.m  # length of the ladder, here the hypotenuse
theta_qty = np.radians(75.0)  # angle to the horizontal

# adj: distance to the building, hyp: length of the ladder
# From eq: cos(theta) =  adj / hyp ==> adj = hyp * cos(theta)
d_qty = l_qty * np.cos(theta_qty)

print(f"horizontal distance: {d_qty:.3g}")
```

A high fountain of water is located at the center of a circular

wet, a student walks around the pool and measures its circumference
to be 15.0 m. Next, the student stands at the edge of

at the bottom of the fountain to be 55.0°. How high is the
fountain?

#### 52. 

A woman measures the angle of elevation of a mountaintop
as 12.0°. After walking 1.00 km closer to the mountain
on level ground, she finds the angle to be 14.0°. Find the
mountain’s height, neglecting the height of the woman’s eyes
above the ground. Hint: Distances from the mountain (x and
x - 1 km) and the mountain’s height are unknown. Draw two
triangles, one for each of the woman’s locations, and equate
expressions for the mountain’s height. Use that expression to
find the first distance x from the mountain and substitute to
find the mountain’s height.

```python
# Define symbolic variables
x, h, theta1, theta2, d = sp.symbols("x h theta1 theta2 d", real=True, positive=True)

# Define trigonometric relations from the two right triangles:
# tan(theta1) = h / x
# tan(theta2) = h / (x - d)
eq1 = sp.Eq(sp.tan(theta1), h / x)
eq2 = sp.Eq(sp.tan(theta2), h / (x - d))

# Solve system of equations for x and h simultaneously
solutions = sp.solve([eq1, eq2], (x, h))

# Simplify expressions using trigonometric identities
x_expr = sp.simplify(solutions[x])
h_expr = sp.simplify(solutions[h])

print("--- Symbolic Formulas ---")
print(f"x = {x_expr}")
print(f"h = {h_expr}")
```

```python
# --- 1. Derive callables from SymPy ---
calc_x = sp.lambdify((d, theta2, theta1), x_expr)
calc_h = sp.lambdify((d, theta1, theta2), h_expr)

# --- 2. Astropy Quantities (_qty suffix) ---
theta1_qty = 12.0 * u.deg
theta2_qty = 14.0 * u.deg
d_qty = 1.0 * u.km

# --- 3. calculations ---
x_qty = calc_x(d_qty, theta2_qty, theta1_qty)
h_qty = calc_h(d_qty, theta1_qty, theta2_qty)

print(f"Mountain height =  {h_qty:.3g} equivalent to {h_qty.to(u.m):.4g}")
```
