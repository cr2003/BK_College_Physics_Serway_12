### 1.2 Measurement Uncertainty and Estimation

#### 31. 
The edges of a shoebox are measured to be 11.4 cm, 17.8 cm,
and 29 cm. Determine the volume of the box, retaining the
proper number of significant figures in your answer.


```python
# Measured values
l_qty = 11.4 * u.cm
w_qty = 17.8 * u.cm
h_qty = 29 * u.cm

# Compute exact volume
exact_vol_qty = l_qty * w_qty * h_qty

# Determine precision: 29 has the less significant figures: 2

print(f"Exact volume = {exact_vol_qty}")
print(f"Final answer (rounded for sig figs) = {exact_vol_qty:.2g}")
```

#### 32. 

Estimate the number of steps you would have to take to walk a
distance equal to the circumference of the Earth.

```python
r_earth_qty = const.R_earth

c_earth_qty = 2 * np.pi * r_earth

# --- assumption: standard human step 70 cm
l_steep_qty = 0.7 * u.m
n_steeps_qty = c_earth_qty / l_steep_qty

print(f"Calculated number of steps: {n_steeps_qty:.3g}")
print(f"Order of magnitude: {oom(n_steeps_qty):.0g}")
```

#### 33. 
**BIO** Estimate the number of breaths taken by a human
being during an average lifetime.

```python
# --- assumptions
# standard adult breath 15 per minute
# human life expectncy 80 years

# 1. Define quantities with AstroPy
breathing_rate_qty = 10 / u.min
lifespan_qty = 70 * u.yr

total_breaths_qty = breathing_rate_qty * lifespan_qty

# Simplify units automatically
total_breaths_qty = total_breaths_qty.decompose()

oom(total_breaths_qty)
print(f"Breathing rate:      {breathing_rate_qty}")
print(f"Lifespan:            {lifespan_qty}")
print(f"Calculated quantity: {total_breaths_qty:.4g}")
print(f"Orden of magnitude:  {oom(total_breaths_qty):.2g}")
```
