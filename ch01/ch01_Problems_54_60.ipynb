#### 54.

Vector $\vec{A}$ has a magnitude of 8.00 units and makes an angle of
45.0° with the positive x-axis. Vector $\vec{B}$
also has a magnitude of 8.00 units and is directed along the negative x-axis. Using
graphical methods, find 

**(a)** the vector sum $\vec{A} + \vec{B}$ and 

**(b)** the vector difference  $\vec{A} - \vec{B}$


```python
# Using Vpython
A_mag = 8.00
theta = vp.radians(45.0)

A = vp.vec(A_mag * vp.cos(theta), A_mag * vp.sin(theta), 0)
B = vp.vec(-8.00, 0, 0)

A_plus_B = A + B
A_minus_B = A - B


print(f"A + B: {fmt_vec(A_plus_B)}")
print(f"A - B: {fmt_vec(A_minus_B)}")
```

#### 55. 

Vector $\vec{A}$ has a magnitude of 29 units and points in the positive
y-direction. When vector $\vec{B}$ is added to $\vec{A}$, the resultant vector
$\vec{A} + \vec{B}$ points in the negative y-direction
with a magnitude of 14 units. Find the magnitude and direction of $\vec{B}$.

```python
A = vp.vec(0, 29, 0)
A_plus_B = vp.vec(0, -14, 0)

B = A_plus_B - A

direction = vp.degrees(vp.atan2(B.y, B.x))

print(f"Vector B<{B.mag}, {direction}°>")
```

An airplane flies $2.00 \times 10^2 km $ due west from city A to
city B and then $3.00 \times 10^2 km $ in the direction of 30.0° north
of west from city B to city C. 

**(a)** In straight-line distance,
how far is city C from city A? 

**(b)** Relative to city A, in what direction is city C? 

**(c)** Why is the answer only approximately

```python
# Import your helper function directly
from helper_functions import math_bearing

AB = vp.vec(-200, 0, 0)
r = 300
theta = vp.radians(180 - 30)
BC = vp.vec(r * vp.cos(theta), r * vp.sin(theta), 0)

AC = AB + BC

theta_AC = vp.degrees(vp.atan2(AC.y, AC.x))
# Convert to navigation bearing
bearing = math_bearing(theta_AC)

print("--- Part (a) ---")
print(f"Distance C from A: {AC.mag:.3g} km")

print("\n--- Part (b) ---")
print(f"Direction from A to C: {theta_AC:.1f}° (Math: from +x-axis)")
print(f"Direction from A to C: {bearing:.1f}° (navigation bearing)")

print("\n--- Part (c) ---")
print(
    "The answer is only approximately correct because the calculation assumes a flat (Euclidean) 2D plane geometry,"
)
print(
    "neglecting the curvature of Earth's spherical surface over these distances (d ≈ 500 km)"
)
```

#### 59. 

A roller coaster moves $ 2.00 \times 10^2 ft$ horizontally and
then rises 135 ft at an angle of 30.0° above the horizontal.
Next, it travels 135 ft at an angle of 40.0° below the horizontal.
Use graphical techniques to find the roller coaster’s
displacement from its starting point to the end of this
movement.

```python
d1 = vp.vec(200, 0, 0)  # ft

d2_mag = 135  # ft
theta_d2 = vp.radians(30.0)
d2 = vp.vec(d2_mag * vp.cos(theta_d2), d2_mag * vp.sin(theta_d2), 0)

d3_mag = 135  # ft
theta_d3 = vp.radians(360 - 40)
d3 = vp.vec(d3_mag * vp.cos(theta_d3), d3_mag * vp.sin(theta_d3), 0)

R = d1 + d2 + d3


theta_R = vp.degrees(vp.atan2(R.y, R.x))
print(f"Resultand force R<{R.mag:.0f}, {theta_R:.0f}°>")
```

Plot the vectors:

```python
# Import your helper function directly
from helper_functions import plot_vector, plot_vector_xy


fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect("equal")
ax.axhline(0, color="gray", linewidth=0.5)
ax.axvline(0, color="gray", linewidth=0.5)
ax.grid(True, linestyle="--", alpha=0.5)

# just in case is needed we can use ret, normally not used
ret_1 = plot_vector_xy(200, 0, ax=ax, color="b", label="d1(200, 0)")
ret_2 = plot_vector(
    135, 30, origin=(d1.x, d1.y), ax=ax, color="g", label="d2(135, 30°)"
)
ret_3 = plot_vector(
    135,
    360 - 40,
    origin=(d1.x + d2.x, d1.y + d2.y),
    ax=ax,
    color="c",
    label="d3(135, 360-40°)",
)
ret_4 = plot_vector_xy(420.33, -19.28, ax=ax, color="r", label="R(200, 0)")


plt.title("Vectores")
plt.show()
```
