#### 11. 
A firkin is an old British unit of volume equal to 9.00 gallons.
How many cubic meters are there in 6.00 firkins?

```python
# 1. Define the custom unit: 1 firkin = 9 gallons
frk = u.def_unit("firkin", 9.00 * u.imperial.gallon)

v_frk_qty = 6.00 * frk

v_gal_qty = v_frk_qty.to(u.m**3)

print(f"{v_frk_qty} = {v_gal_qty:.3g}")
```

#### 13. 
The league, variously defined, is the distance a person can
travel by foot in an hour. It’s still used in parts of rural Mexico
and Latin America. Convert 30.0 m/s to leagues per hour,
assuming 1.00 league = 3.00 mi.

```python
# 1. Define the custom unit: 1 league = 3.00 mi
lg = u.def_unit("league", 3.00 * u.imperial.mi)

sp_qty = 30 * (u.m / u.s)
sp_mph_qty = sp_qty.to(lg / u.h)

print(f"Speed: {sp_qty:.3g} = {sp_mph_qty:.3g}")
```

```python
eff = 25.0 * (u.imperial.mi / u.imperial.gallon)
print(f"Efficiency: {eff:.3g} = {eff.to(u.km / u.l):.3g}")
```

#### 15. 

The diameter of a sphere is measured to be 5.36 in. Find

**(a)** the radius of the sphere in centimeters, 

**(b)** the surface area of the sphere in square centimeters, and 

**(c)** the volume of the sphere in cubic centimeters.

```python
d_qty = 5.36 * u.imperial.inch

# --- Part (a) ---
r_qyt = (d_qty / 2).to(u.cm)

# --- Part (b) ---
# SA = 4 pi r^2
SA_qyt = 4 * np.pi * r_qyt**2

# --- Part (v) ---
# V = 4/3 pi r^3
V_qyt = (4 / 3) * np.pi * r_qyt**3
# --- print results ---
print("\n--- Part (a) ---")
print(f"Sphere radius: {r_qyt:.3g}")

print("\n--- Part (b) ---")
print(f"Sphere Surface Area: {SA_qyt:.3g}")

print("\n--- Part (c) ---")
print(f"Sphere Volume: {V_qyt:.3g}")
```

#### 16. 
**BIO** Suppose your hair grows at the rate of 1/32
inch per day. Find the rate at which it grows in nanometers
per second. Because the distance between atoms in
a molecule is on the order of 0.1 nm, your answer suggests
how rapidly atoms are assembled in this protein
synthesis.

```python
growth_inpday_qty = (1 / 32) * (u.imperial.inch / u.day)
growth_nmps_qty = growth_inpday_qty.to(u.nm / u.s)


#  --- Define the custom unit: 1 atm_layer = 0.1 nm ---
atm_layer = u.def_unit("atom_layers", 0.1 * u.nm)

# --- convert the growth rate to atom_layers per second
rate_qty = growth_nmps_qty.to(atm_layer / u.s)


print(f"Growth rate: {growth_nmps_qty:.2f}")
print(f"Protein synthesis adds layers of atoms at a rate: {rate_qty:.0f}")
```

#### 17. 
The speed of light is about $3.00 \times 10^8 m/s$. Convert this figure
to miles per hour.

```python
C_exact_qyt = const.c
C_approx_qyt = 3e8 * (u.m / u.s)

print(
    f"Exact speed of ligtht: {C_exact_qyt:.0f} = {C_exact_qyt.to(u.imperial.mi / u.h):.0f}"
)
print(
    f"Approx speed of ligtht: {C_approx_qyt:.3g} = {C_approx_qyt.to(u.imperial.mi / u.h):.3g}"
)
```

#### 18. 
A house is 50.0 ft long and 26 ft wide and has 8.0-ft-high
ceilings. What is the volume of the interior of the house in
cubic meters and in cubic centimeters?

```python
l_qty = 50.0 * u.imperial.foot
w_qty = 26 * u.imperial.foot
h_qty = 8.0 * u.imperial.foot

V_qyt = l_qty * w_qty * h_qty

print(f"Volume: {V_qyt.to(u.m**3):.2g} = {V_qyt.to(u.cm**3):.2g}")
```

#### 19. 

The amount of water in reservoirs is often measured in
acre-ft. One acre-ft is a volume that covers an area of one
acre to a depth of one foot. An acre is 43 560 ft2. Find the
volume in SI units of a reservoir containing 25.0 acre-ft of
water.

```python
V_qty = 25.0 * (u.imperial.acre * u.imperial.ft)

print(f"Volume: {V_qty.to(u.m**3):.3g}")
```
