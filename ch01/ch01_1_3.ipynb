### Example 1.8 Number of Galaxies in the Universe

page 11

**Goal** Estimate a volume and a number density, and combine them.

**Problem** Given that astronomers can see about 10 billion light-years into space
and that there are 14 galaxies in our local group, 2 million light-years from the
next local group, estimate the number of galaxies in the observable universe. (Note:
One light-year is the distance traveled by light in one year, about $9.5 \times 10^{15}$ m.)
(See Fig. 1.3.)

<img src="..//pics/Fig_1_3.png" width=400> 

**Strategy** From the known information, we can estimate the number of galaxies per
unit volume. The local group of 14 galaxies is contained in a sphere a million lightyears
in radius, with the Andromeda group in a similar sphere, so there are about
10 galaxies within a volume of radius 1 million light-years. Multiply that number density
by the volume of the observable universe.


```python
### --- Astropy Units ---
glx = u.def_unit("galaxies")

r_lg_qty = 2e6 * u.lyr  # radius of our local group
ng_lg_qty = 14 * glx  # number of galaxies
r_u_qyt = 10e9 * u.lyr  # 10 american billions


# --- calculations ---
# --- volume of local group ---
v_lg_qty = (4 / 3) * np.pi * r_lg_qty**3

# --- estimate the density of galaxies
rho_glx_qty = ng_lg_qty / v_lg_qty

### --- compute the approx volume of observable universe ---
V_u_qyt = (4 / 3) * np.pi * r_u_qyt**3

### --- compute the approx number of galaxies in the observable universe ---
ng_u_qyt = rho_glx_qty * V_u_qyt

### --- print results ---
print(f"Exact calculation: {ng_u_qyt:.3g}")
print(f"Order of magnitude: {oom(ng_u_qyt):.3g}")
```

### Example 1.9 Cartesian and Polar Coordinates

**Goal** Understand how to convert from plane rectangular
coordinates to plane polar coordinates and vice
versa.

**Problem** 

<img src="..//pics/Fig_1_7.png" width=600> 


**(a)** The Cartesian coordinates of a point in
the x y - plane are (x, y) = (-3.50 m, -2.50 m), as shown
in Figure 1.7. Find the polar coordinates of this point.

**Strategy** Apply the trigonometric functions and their inverses, together with the Pythagorean theorem.

```python
# --- (a): Cartesian to polar conversion
x_qty = -3.50 * u.m
y_qty = -2.50 * u.m

# --- calculations ---
r_qty = np.sqrt(x_qty**2 + y_qty**2)
theta_qty = np.degrees(np.atan2(y_qty, x_qty))
theta_norm_qty = (
    theta_qty.value % 360
) * u.deg  # normalize the angle to 360 from +x-axys

print("--- Part (a) ---")
print(f"r = {r_qty:.2f}")
print(f"theta = {theta_norm_qty:.0f}")
```

**(b)** Convert $(r, \theta)$ = (5.00 m, 37.0°) to rectangular
coordinates.

```python
r_qty = 5.00 * u.m
theta_deg_qty = 37.0 * u.deg
theta_rad_qty = theta_deg_qty.to(u.rad)

# --- convert to cartesian coordinates
x_qty = r_qty * np.cos(theta_rad_qty)
y_qty = r_qty * np.sin(theta_rad_qty)


print("--- Part (b) ---")
print(f"x = {x_qty:.2f}")
print(f"y = {y_qty:.2f}")
```

### Example 1.10 How High is the Building?

**Goal** Apply basic results of trigonometry.

**Problem** A person measures the height of a building by walking
out a distance of 46.0 m from its base and shining a flashlight beam
toward the top. When the beam is elevated at an angle of 39.0° with
respect to the horizontal, as shown in Figure 1.8, the beam just
strikes the top of the building. 

**(a)** If the flashlight is held at a height
of 2.00 m, find the height of the building. 

**(b)** Calculate the length
of the light beam.
Strategy Refer to the right triangle shown in the figure. We know
the angle, 39.0°, and the length of the side adjacent to it. Because
the height of the building is the side opposite the angle, we can
use the tangent function. With the adjacent and opposite sides
known, we can then find the hypotenuse with the Pythagorean
theorem.

<img src="..//pics/Fig_1_8.png" width=400> 


```python
# --- AstroPy values ---
h_person = 2.00 * u.m
x_qty = 46.0 * u.m
theta_qty = np.radians(39.0)  # convert 39 degrees to radians

# --- Part (a) ---
# from the equation: tan(theta) = y / x ==> delta_y = x tan(theta)
dy_qty = x_qty * np.tan(theta_qty)

# --- calculate the height of the building
h_building = h_person + dy_qty

# --- Part (b) ---
# calculate hypotenuse r ---
r_qty = np.sqrt(x_qty**2 + dy_qty**2)


# --- print results ---

print("\n--- Part (a) ---")
print(f"Calculated delta y: {dy_qty:.1f} ")
print(f"Building height: {h_building:.1f} ")


print("\n--- Part (b) ---")
print(f"Calculated light beam (r): {r_qty:.1f} ")
```

Could the distance traveled by the light beam be found without using the Pythagorean theorem? How?

```python
# from the equation: cos(theta) = r / x ==> r = x / cos(theta)
r_qty = x_qty / np.cos(theta_qty)

print(f"Yes, it could be calculated using the cosine, (r = x / cos(theta).")
print(f"Calculated light beam (r): {r_qty:.1f} ")
```

While standing atop a building 50.0 m tall, you spot a friend standing on a street corner. Using a protractor and
