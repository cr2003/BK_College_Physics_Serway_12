#### 21. 
A quart container of ice cream is to be made in the form of
a cube. What should be the length of a side, in centimeters?
(Use the conversion 1 gallon = 3.786 liters.)

```python
V_qty = (1 / 4) * u.imperial.gallon

V_qty_cm3 = V_qty.to(u.cm**3)

# --- eq: V = l*3 ==> l = V^(1/3)
l = V_qty_cm3 ** (1 / 3)

print(f"Side: {l:.3g}")
```

#### 26. 
Use your calculator to determine $(\sqrt{8})^3$ to three significant
figures in two ways: 

**(a)** Find $\sqrt{8}$ to four significant figures;
then cube this number and round to three significant
figures. 

**(b)** Find $\sqrt{8}$ to three significant figures; then cube
this number and round to three significant figures. 

**(c)** Which
answer is more accurate? Explain.

```python
import math

"--- Part (a) ---"
sqrt_8_wide = math.sqrt(8)

"--- Part (a) ---"
# Four Significant Figures
sqrt_8_4sf = float(f"{sqrt_8_wide:.4g}")

# cube
sqrt_8_4sf_cube = sqrt_8_4sf**3
# Three Significant Figures
sqrt_8_4sf_cube_3sf = float(f"{sqrt_8_4sf_cube:.3g}")

"--- Part (b) ---"
# Four Significant Figures
sqrt_8_3sf = float(f"{sqrt_8_wide:.3g}")

# cube
sqrt_8_3sf_cube = sqrt_8_3sf**3
# Three Significant Figures
sqrt_8_3sf_cube_3sf = float(f"{sqrt_8_3sf_cube:.3g}")
# --- print results ---
print("\n--- Part (a) ---")
print(f"Wide sqared root of 8: {sqrt_8_wide}")
print(f"Round to four sf: {sqrt_8_4sf}")
print(f"Wide cube root of 8: {sqrt_8_4sf_cube}")
print(f"Round to three sf: {sqrt_8_4sf_cube_3sf}")

print("\n--- Part (b) ---")
print(f"Wide sqared root of 8: {sqrt_8_wide}")
print(f"Round to three sf: {sqrt_8_3sf}")
print(f"Wide cube root of 8: {sqrt_8_3sf_cube}")
print(f"Round to three sf: {sqrt_8_3sf_cube_3sf}")

print("\n--- Part (c) ---")
print("Answer from Part (a) is more accurate.")
print("Rounding early at intermediate step introduces intermediate rounded error.")
```

#### 29. 
A rectangle has a length of  ($2.0 \pm 0.2$) m and a width of
( $1.5 \pm 0.1$) m. Calculate 

**(a)** the area and 

**(b)** the perimeter of
the rectangle, and give the uncertainty in each value.

```python
# Length and Width inputs
L, dL = 2.0, 0.2
W, dW = 1.5, 0.1

# (a) Area calculation
A = L * W
dA_fractional = A * (dL / L + dW / W)
dA_quadrature = A * np.sqrt((dL / L) ** 2 + (dW / W) ** 2)

# (b) Perimeter calculation
P = 2 * L + 2 * W
dP = 2 * dL + 2 * dW

print(f"Area (Linear/Fractional): {A:.1f} +/- {dA_fractional:.1f} m^2")
print(f"Area (Quadrature):        {A:.1f} +/- {dA_quadrature:.1f} m^2")
print(f"Perimeter:                {P:.1f} +/- {dP:.1f} m")
```

#### 30. 

The radius of a circle is measured to be ( $10.5 \pm 0.2$) m. Calculate

**(a)** the area and 

**(b)** the circumference of the circle, and
give the uncertainty in each value.

```python
r = 10.5
dr = 0.2

# (a) Area and uncertainty
A = np.pi * r**2
dA = 2 * np.pi * r * dr  # derivative of previous A

# (b) Circumference and uncertainty
C = 2 * np.pi * r
dC = 2 * np.pi * dr

print(f"Area:          {A:.1f} +/- {dA:.1f} m^2")
print(f"Circumference: {C:.1f} +/- {dC:.1f} m")
```
